```python
# 计算对应epoch的余弦退火策略下的学习率
import numpy as np

# 定义初始参数
initial_learning_rate = 0.0001
total_epochs = 120
epoch_70 = 70

# 使用余弦退火公式计算学习率
learning_rate_70_epochs = initial_learning_rate * 0.5 * (1 + np.cos(epoch_70 * np.pi / total_epochs))

print(learning_rate_70_epochs)
```
